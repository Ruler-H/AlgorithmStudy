```python
# 문자열 겹쳐쓰기
def solution(my_string, overwrite_string, s):
    return my_string[:s] + overwrite_string + my_string[len(overwrite_string) + s:]

solution("He11oWor1d", "lloWorl", 2)
```

```python
# 문자열 섞기
def solution(str1, str2):
    answer = ''
    for i in range(len(str1)):
        answer += str1[i]
        answer += str2[i]
    return answer
```

```python
# 문자 리스트를 문자열로 변환하기
def solution(arr):
    answer = "".join(arr)
    return answer

solution(["a","b","c"])
```

```python
# 공배수
def solution(number, n, m):
    if number % n == 0 and number % m == 0:
        return 1
    else:
        return 0
```

```python
# 홀짝에 따라 다른 값 반환하기
def solution(n):
    answer = 0
    if n % 2 == 0:
        for i in range(2, n + 1, 2):
            answer += i ** 2
    else:
        for i in range(1, n + 1, 2):
            answer += i

    return answer
```

```python
# 조건 문자열
def solution(ineq, eq, n, m):
    if ineq == "<":
        if eq == "=":
            if n <= m:
                return 1
            return 0
        else:
            if n < m:
                return 1
            return 0
    else:
        if eq == "=":
            if n >= m:
                return 1
            return 0
        else:
            if n > m:
                return 1
            return 0
```

```python
# flag에 따라 다른 값 반환하기
def solution(a, b, flag):
    if flag:
        return a + b
    return a - b
```

```python
# 코드 처리하기
def solution(code):
    ret = ''
    mode = 0
    for i in range(len(code)):
        if code[i] == "1":
            mode = 1 if mode == 0 else 0
        else:
            if mode:
                if i % 2 == 1:
                    ret += code[i]
            else:
                if i % 2 == 0:
                    ret += code[i]
    if ret:
        return ret
    else:
        return "EMPTY"

solution("abc1abc1abc")
```

```python
# 등차수열의 특정한 항만 더하기
def solution(a, d, included):
    answer = 0
    for i in range(len(included)):
        answer += (a + (d * i) if included[i] else 0)

    return answer
```

```python
# 주사위 게임 2
def solution(a, b, c):
    if a != b != c and a != c:
        return a + b + c
    if a == b == c:
        return (a + b + c) * (a ** 2 + b ** 2 + c ** 2) * (a ** 3 + b ** 3 + c ** 3)
    return (a + b + c) * (a ** 2 + b ** 2 + c ** 2)

solution(2, 6, 1), solution(5, 3, 3), solution(4, 4, 4), solution(1, 1, 1)
solution(3, 5, 3)
```

```python
# 이어 붙인 수
def solution(num_list):
    even = ""
    odd = ""
    for i in num_list:
        if i % 2 == 0:
            even += str(i)
        else:
            odd += str(i)
    return int(even) + int(odd)

solution([3, 4, 5, 2, 1])
```
