```python
# 수 조작하기 1
def solution(n, control):
    for i in control:
        if i == "w":
            n += 1
        elif i == "s":
            n -= 1
        elif i == "d":
            n += 10
        elif i == "a":
            n -= 10
    return n
```

```python
# 수 조작하기 2
def solution(numLog):
    answer = ''
    for i in range(1, len(numLog)):
        if numLog[i] - numLog[i - 1] == 1:
            answer += "w"
        elif numLog[i] - numLog[i - 1] == -1:
            answer += "s"
        elif numLog[i] - numLog[i - 1] == 10:
            answer += "d"
        elif numLog[i] - numLog[i - 1] == -10:
            answer += "a"
    return answer
```

```python
# 수열과 구간 쿼리 3
def solution(arr, queries):
    for i, j in queries:
        arr[i], arr[j] = arr[j], arr[i]
    return arr
```

```python
# 수열과 구간 쿼리 2
def solution(arr, queries):
    answer = []
    for s, e, k in queries:
        temp = []
        for i in range(s, e + 1):
            if arr[i] > k:
                temp.append(arr[i])
        if temp:
            answer.append(min(temp))
        else:
            answer.append(-1)
    return answer
```

```python
# 수열과 구간 쿼리 4
def solution(arr, queries):
    for s, e, k in queries:
        for i in range(s, e + 1):
            if i % k == 0:
                arr[i] += 1
    return arr
```

```python
# 배열 만들기 2
def solution(l, r):
    answer = []
    for i in range(l, r + 1):
        if set(str(i)) <= {'0', '5'}:
            answer.append(i)
    return answer if answer else [-1]
```

```python
# 콜라츠 수열 만들기
def solution(n):
    answer = []
    while n != 1:
        answer.append(n)
        if n % 2 == 0:
            n = n // 2
        else:
            n = 3 * n + 1
    answer.append(1)
    return answer
```

```python
# 배열 만들기 4
def solution(arr):
    stk = []
    i = 0
    while i < len(arr):
        if stk:
            if stk[-1] < arr[i]:
                stk.append(arr[i])
                i += 1
            else:
                stk.pop()
        else:
            stk.append(arr[i])
            i += 1
    return stk

solution([1, 4, 2, 5, 3])
```

```python
# 글자 이어 붙여 문자열 만들기
def solution(my_string, index_list):
    answer = ''
    for i in index_list:
        answer += my_string[i]
    return answer
solution("zpiaz", [1, 2, 0, 0, 3])
```

```python
# 9로 나눈 나머지
def solution(number):
    num_sum = sum(int(i) for i in number)
    return num_sum % 9
```

```python
# 문자열 여러 번 뒤집기
def solution(my_string, queries):
    for s, e in queries:
        my_string = my_string[:s] + my_string[s:e+1][::-1] + my_string[e+1:]
    return my_string
solution("rermgorpsam", [[2, 3], [0, 7], [5, 9], [6, 10]])
```

```python
# 배열 만들기 5
def solution(intStrs, k, s, l):
    answer = []
    for i in intStrs:
        if int(i[s:s+l]) > k:
            answer.append(int(i[s:s+l]))
    return answer
```

```python
# 부분 문자열 이어 붙여 문자열 만들기
def solution(my_strings, parts):
    answer = ''
    for i in range(len(my_strings)):
        answer += my_strings[i][parts[i][0]:parts[i][1]+1]
    return answer
solution(["progressive", "hamburger", "hammer", "ahocorasick"], [[0, 4], [1, 2], [3, 5], [7, 7]])
```
