```python
# n보다 커질 때까지 더하기
def solution(numbers, n):
    answer = 0
    for number in numbers:
        if answer > n:
            return answer
        answer += number
    return answer
```

```python
# 수열과 구간 쿼리 1
def solution(arr, queries):
    for s, e in queries:
        for i in range(s, e + 1):
            arr[i] += 1
    return arr
```

```python
# 조건에 맞게 수열 변환하기 1
def solution(arr):
    for i in range(len(arr)):
        if arr[i] >= 50 and arr[i] % 2 == 0:
            arr[i] //= 2
        elif arr[i] < 50 and arr[i] % 2 == 1:
            arr[i] *= 2
    return arr
```

```python
# 조건에 맞게 수열 변환하기 2
def solution(arr):
    cnt = 0
    while True:
        new_arr = []
        for i in range(len(arr)):
            if arr[i] >= 50 and arr[i] % 2 == 0:
                new_arr.append(arr[i] // 2)
            elif arr[i] < 50 and arr[i] % 2 == 1:
                new_arr.append(arr[i] * 2 + 1)
            else:
                new_arr.append(arr[i])
        
        for i in range(len(arr)):
            if arr[i] != new_arr[i]:
                print(arr[i], new_arr[i])
                break
        else:
            return cnt
        arr = new_arr
        cnt += 1

solution([1, 2, 3, 100, 99, 98])
```

```python
# 1로 만들기
def solution(num_list):
    answer = 0
    for i in num_list:
        while i != 1:
            answer += 1
            if i % 2 == 0:
                i //= 2
            else:
                i = (i - 1) // 2

    return answer

solution([12, 4, 15, 1, 14])
```
