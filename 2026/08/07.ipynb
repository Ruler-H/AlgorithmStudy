```python
# 길이에 따른 연산
import math
def solution(num_list):
    if len(num_list) >= 11:
        return sum(num_list)
    return math.prod(num_list)

solution([3, 4, 5, 2, 5, 4, 6, 7, 3, 7, 2, 2, 1]), solution([2, 3, 4, 5])
```

```python
# 원하는 문자열 찾기
def solution(myString, pat):
    myString = myString.upper()
    return 1 if pat.upper() in myString else 0

solution("AbCdEfG", "aBc")
```

```python
# 배열에서 문자열 대소문자 변환하기
def solution(strArr):
    for i in range(len(strArr)):
        if i % 2 == 0:
            strArr[i] = strArr[i].lower()
        else:
            strArr[i] = strArr[i].upper()
    return strArr
```

```python
# A 강조하기
def solution(myString):
    answer = ""
    for i in range(len(myString)):
        if myString[i] in ["A", "a"]:
            answer += "A"
        else:
            answer += myString[i].lower()
    return answer

solution("abstract algebra")
```

```python
# 특정한 문자를 대문자로 바꾸기
def solution(my_string, alp):
    return my_string.replace(alp, alp.upper())
```

```python
# 특정 문자열로 끝나는 가장 긴 부분 문자열 찾기
def solution(myString, pat):
    for i in range(len(myString), -1, -1):
        if myString[i - len(pat):i] == pat:
            return myString[:i]
solution("AbCdEFG", "dE")
```

```python
# 문자열이 몇 번 등장하는지 세기
def solution(myString:str, pat):
    cnt = 0
    for i in range(len(myString)):
        if myString[i:i+len(pat)] == pat:
            cnt += 1
    return cnt

solution("banana", "ana")
```

```python
# ad 제거하기
def solution(strArr):
    answer = []
    for i in range(len(strArr)):
        if "ad" not in strArr[i]:
            answer.append(strArr[i])
    return answer

solution(["and","notad","abcd"])
```

```python
# 공백으로 구분하기 2
def solution(my_string):
    answer = []
    temp = ""
    for i in my_string:
        if i != " ":
            temp += i
        else:
            if temp:
                answer.append(temp)
                temp = ""
    if temp:
        answer.append(temp)
    return answer

solution(" i    love  you")
```

```python
# x 사이의 개수
def solution(myString):
    return [len(i) for i in myString.split("x")]

solution("oxooxoxxox")
```

```python
# 문자열 잘라서 정렬하기
def solution(myString):
    str_list = myString.split("x")
    str_list.sort()
    for i in range(len(str_list)):
        if str_list[i]:
            return str_list[i:]

solution("dxccxbbbxaaaa")
```

```python
# 간단한 식 계산하기
def solution(binomial):
    a, op, b = binomial.split(" ")
    a, b = int(a), int(b)
    if op == "*":
        return a * b
    elif op == "+":
        return a + b
    elif op == "-":
        return a - b

solution("43 + 12")
```

```python
# 문자열 바꿔서 찾기
def solution(myString, pat):
    repl_str = ""
    for i in myString:
        repl_str += "B" if i == "A" else "A"
    return 1 if pat in repl_str else 0
```

```python
# 세 개의 구분자
def solution(myStr):
    answer = []
    for i in myStr.replace("b", "a").replace("c", "a").split("a"):
        if i:
            answer.append(i)
    if not answer:
        answer.append("EMPTY")
    return answer

solution("baconlettucetomato")
```

```python
# 배열의 원소만큼 추가하기
def solution(arr):
    answer = []
    for i in arr:
        answer.extend([i] * i)
    return answer

solution([5, 1, 4])
```
