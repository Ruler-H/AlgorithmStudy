```python
# 빈 배열에 추가, 삭제하기
def solution(arr, flag):
    answer = []
    for i in range(len(arr)):
        if flag[i]:
            answer.extend([arr[i]] * (arr[i] * 2))
        else:
            answer = answer[:len(answer) - arr[i]]
        
    return answer

solution([3, 2, 4, 1, 3], [True, False, True, False, False]	)
```

```python
# 배열 만들기 6
def solution(arr):
    answer = []
    for i in range(len(arr)):
        if not answer:
            answer.append(arr[i])
        else:
            if answer[-1] == arr[i]:
                answer = answer[:-1]
            else:
                answer.append(arr[i])
    return answer if answer else [-1]
```

```python
# 무작위로 K개의 수 뽑기
def solution(arr, k):
    answer = []
    for i in arr:
        if i in answer:
            continue
        else:
            answer.append(i)
        if len(answer) == k:
            return answer
    answer.extend([-1] * (k - len(answer)))
    return answer
```

```python
# 배열의 길이를 2의 거듭제곱으로 만들기
def solution(arr):
    target = 0
    for i in range(0, 11):
        if len(arr) <= (2 ** i):
            target = 2 ** i
            break
    arr.extend([0] * (target - len(arr)))
    return arr
```

```python
# 배열 비교하기
def solution(arr1, arr2):
    if len(arr1) > len(arr2):
        return 1
    elif len(arr1) < len(arr2):
        return -1
    else:
        if sum(arr1) > sum(arr2):
            return 1
        elif sum(arr1) < sum(arr2):
            return -1
        else:
            return 0
```

```python
# 문자열 묶기
def solution(strArr):
    result = [0] * 31
    for i in strArr:
        result[len(i)] += 1

    return max(result)
```

```python
# 배열의 길이에 따라 다른 연산하기
def solution(arr, n):
    start_idx = 1
    if len(arr) % 2 == 1:
        start_idx = 0
    for i in range(start_idx, len(arr), 2):
        arr[i] += n
    return arr
```

```python
# 전국 대회 선발 고사
def solution(rank:list, attendance):
    answer = 0
    find_rank = 1
    for i in range(1, len(rank) + 1):
        std_idx = rank.index(i)
        if attendance[std_idx]:
            if find_rank == 1:
                answer += (10000 * std_idx)
            elif find_rank == 2:
                answer += (100 * std_idx)
            else:
                answer += std_idx
                break
            find_rank += 1
    return answer

solution([3, 7, 2, 5, 4, 6, 1], [False, True, True, True, True, False, False])
```

```python
# 문자열 정수의 합
def solution(num_str):
    answer = 0
    for i in num_str:
        answer += int(i)
    return answer
```

```python
# 배열의 원소 삭제하기
def solution(arr:list, delete_list):
    for i in delete_list:
        if i in arr:
            arr.remove(i)
    return arr

solution([293, 1000, 395, 678, 94], [94, 777, 104, 1000, 1, 12])
```

```python
# 꼬리 문자열
def solution(str_list, ex):
    answer = ''
    for i in str_list:
        if ex not in i:
            answer += i
    return answer
```
