```python
# 날짜 비교하기
from datetime import datetime
def solution(date1, date2):
    date1 = datetime(*date1)
    date2 = datetime(*date2)
    if date1 < date2:
        return 1
    else:
        return 0

solution([2021, 12, 28], [2021, 12, 29])
```

```python
# 그림 확대
def solution(picture, k):
    answer = []
    for i in picture:
        result = ""
        for j in i:
            result += (j * k)
        answer.extend([result for i in range(k)])
    return answer

solution([".xx...xx.", "x..x.x..x", "x...x...x", ".x.....x.", "..x...x..", "...x.x...", "....x...."], 2)
```

```python
# 조건에 맞게 수열 변환하기 3
def solution(arr, k):
    if k % 2 == 1:
        for i in range(len(arr)):
            arr[i] *= k
    else:
        for i in range(len(arr)):
            arr[i] += k
    return arr
```

```python
# l로 만들기
def solution(myString):
    after = ['m', 'n', 'o', 'p', 'q', 'r', 's', 't', 'u', 'v', 'w', 'x', 'y', 'z']
    answer = ""
    for i in myString:
        if i not in after:
            answer += "l"
        else:
            answer += i
    return answer
```

```python
# 특별한 이차원 배열 1
def solution(n):
    answer = [[0] * n for i in range(n)]
    for i in range(n):
        answer[i][i] = 1
    return answer

solution(3)
```

```python
# 정수를 나선형으로 배치하기
def solution(n):
    answer = [[0] * n for i in range(n)]
    directions = [(0, 1), (1, 0), (0, -1), (-1, 0)]
    x, y = 0, 0
    d_idx = 0
    for i in range(1, n ** 2 + 1):
        answer[x][y] = i
        next_x, next_y = x + directions[d_idx][0], y + directions[d_idx][1]
        if next_x >= n or next_y >= n or answer[next_x][next_y] != 0:
            d_idx += 1
            d_idx %= 4
        x, y = x + directions[d_idx][0], y + directions[d_idx][1]
    return answer

solution(5)
```

```python
# 특별한 이차원 배열 2
def solution(arr):
    for i in range(len(arr)):
        for j in range(len(arr)):
            if arr[i][j] != arr[j][i]:
                return 0
    return 1
```

```python
# 정사각형으로 만들기
def solution(arr):
    i_len = len(arr)
    j_len = len(arr[0])
    if i_len > j_len:
        for i in range(i_len):
            arr[i].extend([0] * (i_len - j_len))
    else:
        arr.extend([[0] * j_len] * (j_len - i_len))
    return arr

solution([[572, 22, 37], [287, 726, 384], [85, 137, 292], [487, 13, 876]])
```

```python
# 이차원 배열 대각선 순회하기
def solution(board, k):
    answer = 0
    for i in range(len(board)):
        for j in range(len(board[0])):
            if i + j <= k:
                answer += board[i][j]
    return answer

solution([[0, 1, 2],[1, 2, 3],[2, 3, 4],[3, 4, 5]], 2)
```

```python
# 다음에 올 숫자
def solution(common):
    if common[1] - common[0] == common[2] - common[1]:
        return common[-1] + (common[1] - common[0])
    else:
        return common[-1] * (common[1] // common[0])
```

```python
# 연속된 수의 합
def solution(num, total):
    start_num = 0
    if num % 2 == 1:
        start_num = (total // num) - ((num - 1) // 2)
    else:
        start_num = (total // num) - ((num // 2) - 1)
    answer = []
    for i in range(start_num, start_num + num):
        answer.append(i)
    return answer

solution(3, 12)
```

```python
# 종이 자르기
def solution(M, N):
    answer = M - 1
    answer += M * (N - 1)
    return answer
```

```python
# 문자열 밀기
def solution(A, B):
    if A == B:
        return 0
    for i in range(1, len(A)):
        A = A[-1] + A[:-1]
        if A == B:
            return i
    return -1
```
