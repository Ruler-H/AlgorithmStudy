```python
# 7의 개수
def solution(array):
    answer = 0
    for i in array:
        answer += str(i).count("7")
    return answer
```

```python
# 문자열 정렬하기 (2)
def solution(my_string:str):
    return "".join(sorted([i for i in my_string.lower()]))

solution("Bcad")
```

```python
# 제곱수 판별하기
import math
def solution(n):
    root = math.isqrt(n)
    return 1 if n == root ** 2 else 2

solution(25)
```

```python
# 문자열안에 문자열
def solution(str1, str2):
    return 1 if str2 in str1 else 2
```

```python
# OX퀴즈
def solution(quiz):
    answer = []
    for i in quiz:
        x, oper, y, _, z = i.split()
        x, y, z = int(x), int(y), int(z)
        if oper == "+":
            answer.append("O" if x + y == z else "X")
        else:
            answer.append("O" if x - y == z else "X")
    return answer

solution(["3 - 4 = -3", "5 + 6 = 11"])
```

```python
# n의 배수 고르기
def solution(n, numlist):
    return list(filter(lambda x: x % n == 0, numlist))
```

```python
# 숫자 찾기
def solution(num:int, k:int):
    num, k = str(num), str(k)
    k_idx = num.find(k)
    return -1 if k_idx == -1 else k_idx + 1

solution(29183, 1), solution(232443, 4), solution(123456, 7)
```

```python
# 배열의 유사도
def solution(s1, s2):
    answer = 0
    for i in s1:
        if i in s2:
            answer += 1
    return answer
```

```python
# 문자열 계산하기
def solution(my_string):
    oper_list = my_string.split()
    result = int(oper_list[0])
    for i in range(2, len(oper_list), 2):
        if oper_list[i - 1] == "+":
            result += int(oper_list[i])
        else:
            result -= int(oper_list[i])
    return result

solution("3 + 4")
```

```python
# 가장 큰 수 찾기
def solution(array:list):
    max_num = max(array)
    max_idx = array.index(max_num)
    return [max_num, max_idx]
```

```python
# 약수 구하기
def solution(n):
    answer = []
    for i in range(1, int(n ** 0.5) + 1):
        if n % i == 0:
            answer.append(i)
            if i != n // i:
                answer.append(n // i)
    answer.sort()
    return answer

solution(24)
```

```python
# 한 번만 등장한 문자
def solution(s):
    if len(s) == 1:
        return s
    answer = ''
    s_list = sorted(list(map(lambda x: x, s)))
    if s_list[0] != s_list[1]:
        answer += s_list[0]
    for i in range(1, len(s_list) - 1):
        if s_list[i - 1] != s_list[i] and s_list[i] != s_list[i + 1]:
            answer += s_list[i]
    if s_list[-2] != s_list[-1]:
        answer += s_list[-1]
    return answer

solution("abcabcadc"), solution("abdc"), solution("hello")
```
