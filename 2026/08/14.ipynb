```python
# 인덱스 바꾸기
def solution(my_string, num1, num2):
    my_string_list = list(map(lambda x: x, my_string))
    my_string_list[num1], my_string_list[num2] = my_string_list[num2], my_string_list[num1]
    return "".join(my_string_list)

solution("hello", 1, 2)
```

```python
# 대문자와 소문자
def solution(my_string:str):
    answer = ''
    for i in my_string:
        if i.isupper():
            answer += i.lower()
        else:
            answer += i.upper()
    return answer
```

```python
# 암호 해독
def solution(cipher, code):
    answer = ''
    for i in range(code - 1, len(cipher), code):
        answer += cipher[i]
    return answer

solution("dfjardstddetckdaccccdegk", 4)
```

```python
# 가까운 수
def solution(array, n):
    answer = array[0]
    for i in range(1, len(array)):
        if abs(array[i] - n) < abs(answer - n) or (abs(array[i] - n) == abs(answer - n) and array[i] < answer):
            answer = array[i]
        
    return answer
```

```python
# 삼각형의 완성조건 (1)
def solution(sides):
    max_len = max(sides)
    max_len_idx = sides.index(max_len)
    left_len_sum = 0
    for i in range(3):
        if max_len_idx != i:
            left_len_sum += sides[i]
    return 1 if max_len < left_len_sum else 2

solution([1, 2, 3]), solution([3, 6, 2]), solution([199, 72, 222]), solution([1, 1, 1])
```

```python
# 중복된 문자 제거
def solution(my_string):
    answer = ""
    for i in my_string:
        if i not in answer:
            answer += i
    return answer

solution("people")
```

```python
# k의 개수
def solution(i, j, k):
    answer = 0
    for num in range(i, j + 1):
        answer += str(num).count(str(k))
    return answer

solution(1, 13, 1)
```

```python
# A로 B 만들기
def solution(before, after):
    for i in range(len(before)):
        if before.count(before[i]) != after.count(before[i]):
            return 0
    return 1
```

```python
# 치킨 쿠폰
def solution(chicken):
    answer = 0
    while chicken >= 10:
        service = chicken // 10
        answer += service
        chicken = (chicken % 10) + service
    return answer

solution(100)
```

```python
# 로그인 성공?
def solution(id_pw, db):
    if id_pw in db:
        return "login"
    for id, pw in db:
        if id_pw[0] == id and id_pw[1] != pw:
            return "wrong pw"
    return "fail"

solution(["meosseugi", "1234"], [["rardss", "123"], ["yyoom", "1234"], ["meosseugi", "1234"]])
```

```python
# 등수 매기기
def solution(score):
    score_sum_list = sorted(map(lambda x: x[0] + x[1], score), reverse=True)
    return [score_sum_list.index(x + y) + 1 for x, y in score]

solution([[80, 70], [90, 50], [40, 70], [50, 80]])
```

```python
# 특이한 정렬
def solution(numlist:list, n):
    numlist.sort(key=lambda x: (abs(x - n), -x))
    return numlist

solution([1, 2, 3, 4, 5, 6], 4)
```

```python
# 유한소수 판별하기
def solution(a, b):
    for i in range(2, min(a, b) +  1):
        if a % i == 0 and b % i == 0:
            a //= i
            b //= i
    while b % 2 == 0:
        b //= 2
    while b % 5 == 0:
        b //= 5
    return 1 if b == 1 else 2

solution(11, 22)
```

```python
# 겹치는 선분의 길이
def solution(lines):
    total_line = [0] * 200
    for start, end in lines:
        for  i in range(start, end):
            total_line[i] += 1
    return len(list(filter(lambda x: x >= 2, total_line)))

solution([[0, 1], [2, 5], [3, 9]]), solution([[-1, 1], [1, 3], [3, 9]]), solution([[0, 5], [3, 9], [1, 10]])
```

```python
# 평행
def solution(dots):
    x1, y1 = dots[0]
    x2, y2 = dots[1]
    x3, y3 = dots[2]
    x4, y4 = dots[3]
    if ((y1 - y2) / (x1 - x2) == (y3 - y4) / (x3 - x4)) or ((y1 - y3) / (x1 - x3) == (y2 - y4) / (x2 - x4)) or ((y1 - y4) / (x1 - x4) == (y2- y3) / (x2 - x3)):
        return 1
    return 0

solution([[1, 4], [9, 2], [3, 8], [11, 6]]), solution([[3, 5], [4, 1], [2, 4], [5, 10]])
```
