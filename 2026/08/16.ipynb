```python
# 캐릭터의 좌표
def solution(keyinput, board):
    x, y = 0, 0
    x_min_len = -((board[0] - 1) // 2)
    x_max_len = ((board[0] - 1) // 2)
    y_min_len = -((board[1] - 1) // 2)
    y_max_len = ((board[1] - 1) // 2)
    direction = {
        "up": (0, 1),
        "down": (0, -1),
        "left": (-1, 0),
        "right": (1, 0)
    }
    for i in keyinput:
        x_move, y_move = direction[i]
        if x_min_len <= x + x_move <= x_max_len:
            x += x_move
        if y_min_len <= y + y_move <= y_max_len:
            y += y_move
    return [x, y]
```

```python
# 직사각형 넓이 구하기
def solution(dots):
    x_len, y_len = 0, 0
    for i in dots[1:]:
        if dots[0][0] == i[0]:
            y_len = abs(dots[0][1] - i[1])
        elif dots[0][1] == i[1]:
            x_len = abs(dots[0][0] - i[0])
    return x_len * y_len
```

```python
# 컨트롤 제트
def solution(s:str):
    s_list = s.split(" ")
    answer = 0
    while s_list:
        s_str = s_list.pop()
        if s_str == "Z":
            s_list.pop()
        else:
            answer += int(s_str)
    return answer
```

```python
# 소인수분해
def solution(n):
    answer = []
    i = 2
    while n != 1:
        if n % i == 0:
            answer.append(i)
            while n % i == 0:
                n //= i
        i += 1
        
    return answer

solution(12), solution(17), solution(420)
```

```python
# 문자열 정렬하기 (1)
def solution(my_string:str):
    filtered_my_string = filter(lambda x: x.isnumeric() , my_string)
    int_list = sorted(list(map(lambda x: int(x), filtered_my_string)))
    return int_list

solution("hi12392")
```

```python
# 팩토리얼
def solution(n):
    mul_num = 1
    i = 2
    while True:
        mul_num *= i
        if n == mul_num:
            return i
        elif n < mul_num:
            return i - 1
        i += 1
```

```python
# 합성수 찾기
def solution(n):
    answer = 0
    for i in range(2, n + 1):
        for j in range(2, i):
            if i % j == 0:
                answer += 1
                break
    return answer

solution(10), solution(15)
```

```python
# 주사위의 개수
def solution(box, n):
    return (box[0] // n) * (box[1] // n) * (box[2] // n)
```

```python
# 공 던지기
def solution(numbers, k):
    ball_idx = (2 * (k - 1)) % len(numbers)
    return numbers[ball_idx]

solution([1, 2, 3, 4], 2)
```

```python
# 2차원으로 만들기
def solution(num_list, n):
    answer = []
    temp = []
    while num_list:
        temp.append(num_list.pop(0))
        if len(temp) == n:
            answer.append(temp)
            temp = []
    return answer
```

```python
# 구슬을 나누는 경우의 수
def solution(balls, share):
    answer = 1
    for i in range(balls, balls - share, -1):
        answer *= i
    for i in range(2, 1 + share):
        answer //= i
    return answer

solution(3, 2)
```
