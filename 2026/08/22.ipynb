```python
# 노란불 신호등
import math

def solution(signals):
    max_val = math.lcm(*map(lambda x : sum(x), signals))
    answer = -1
    for i in range(1, max_val + 1):
        flag = True
        for signal in signals:
            if not signal[0] < (i % sum(signal)) <= sum(signal[:2]):
                flag = False
                break
        if flag:
            answer = i
            break
    return answer

solution([[2, 1, 2], [5, 1, 1]]), solution([[2, 3, 2], [3, 1, 3], [2, 1, 1]]), solution([[3, 3, 3], [5, 4, 2], [2, 1, 2]]), solution([[1, 1, 4], [2, 1, 3], [3, 1, 2], [4, 1, 1]])
```

```python
# 중요한 단어를 스포 방지
def solution(message, spoiler_ranges):
    spoiler_words = set()
    for start, end in spoiler_ranges[::-1]:
        if message[start] != " ":
            while start > 0 and message[start - 1] != " ":
                start -= 1
        if message[end] != " ":
            while end + 1 < len(message) and message[end + 1] != " ":
                end += 1
        spoiler_words = spoiler_words.union(filter(lambda x : x, message[start:end+1].split(" ")))
        masked = "".join("#" if c != " " else " " for c in message[start:end+1])
        message = message[:start] + masked + message[end + 1:]
    answer = 0
    message_list = message.split(" ")
    for spoiler_word in spoiler_words:
        if not spoiler_word in message_list:
            answer += 1
    return answer

# solution("here is muzi here is a secret message", [[0, 3], [23, 28]])
# solution("my phone number is 01012345678 and may i have your phone number", [[5, 5], [25, 28], [34, 40], [53, 59]])
# solution("a ba ba", [[0, 4]])
# solution(message = "bbb aa",  spoiler_ranges = [[0, 1], [2, 3]])
solution(message = "aa ba ba",  spoiler_ranges = [[1, 3], [4, 7]])
```
