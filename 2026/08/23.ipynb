```python
# 택배 상자 꺼내기
def solution(n, w, num):
    box_array = [[0] * w for i in range((n // w) + 1)]
    target_i, target_j = 0, 0
    for i in range(1, n + 1):
        line = ((i - 1) // w)
        if line % 2 == 0:
            x, y = line, (i - 1) % w
        else:
            x, y = line, -(i % w)
        box_array[x][y] = i
        if i == num:
            target_i, target_j = x, y

    if box_array[-1][target_j] != 0:
        return len(box_array) - target_i
    else:
        return len(box_array) - target_i - 1


solution(22, 6, 8), solution(13, 3, 6)
```

```python
# 유연근무제
def solution(schedules, timelogs, startday):
    result_list = [1] * len(schedules)
    if startday != 7:
        for i in timelogs:
            for j in range(2):
                i.pop(6 - startday)
    else:
        for i in timelogs:
                i.pop(0)
                i.pop(-1)
    for i in range(len(schedules)):
        expiration_time = schedules[i] + 10
        if (expiration_time % 100) >= 60:
            expiration_time += 40
        for j in timelogs[i]:
            if j > expiration_time:
                result_list[i] = 0
                break
    return sum(result_list)

solution([700, 800, 1100], [[710, 2359, 1050, 700, 650, 631, 659], [800, 801, 805, 800, 759, 810, 809], [1105, 1001, 1002, 600, 1059, 1001, 1100]], 5)
solution([730, 855, 700, 720], [[710, 700, 650, 735, 700, 931, 912], [908, 901, 805, 815, 800, 831, 835], [705, 701, 702, 705, 710, 710, 711], [707, 731, 859, 913, 934, 931, 905]], 1)
```
