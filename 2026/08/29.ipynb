```python
# [PCCP 기출문제] 1번 / 동영상 재생기
def solution(video_len, pos, op_start, op_end, commands):
    video_len = parseToSeconds(video_len)
    pos = parseToSeconds(pos)
    op_start = parseToSeconds(op_start)
    op_end = parseToSeconds(op_end)
    for i in commands:
        pos = skip_if_in_range(pos, op_start, op_end)
        if i == "next":
            pos += 10
            pos = min(video_len, pos)
        elif i == "prev":
            pos -= 10
            pos = max(0, pos)
    pos = skip_if_in_range(pos, op_start, op_end)
    return formatToTime(pos)

def parseToSeconds(time : str) -> int :
    minute, second = time.split(":")
    return (int(minute) * 60) + (int(second))

def formatToTime(second : int) -> str :
    return f"{second // 60:02d}:{second % 60:02d}"

def skip_if_in_range(seconds : int, start : int, end : int) -> int:
    if start <= seconds <= end:
        return end
    return seconds

solution("34:33", "13:00", "00:55", "02:55", ["next", "prev"]), solution("10:55", "00:05", "00:15", "06:55", ["prev", "next", "next"]), solution("07:22", "04:05", "00:15", "04:07", ["next"])
```
