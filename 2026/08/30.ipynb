```python
# [PCCE 기출문제] 9번 / 지폐 접기
def solution(wallet, bill):
    answer = 0
    while not ((bill[0] <= wallet[0] and bill[1] <= wallet[1]) or (bill[1] <= wallet[0] and bill[0] <= wallet[1])):
        answer += 1
        if bill[0] > bill[1]:
            bill[0] //= 2
        else:
            bill[1] //= 2
    return answer

solution([30, 15], [26, 17]), solution([50, 50], [100, 241])
```

```python
# 가장 많이 받은 선물
def solution(friends, gifts):
    answer = 0
    gifts_map = [[0] * len(friends) for i in range(len(friends))]
    for gift in gifts:
        sender, receiver = gift.split(" ")
        sender_idx, receiver_idx = friends.index(sender), friends.index(receiver)
        gifts_map[sender_idx][receiver_idx] += 1
    gifts_score = []
    for i in range(len(friends)):
        gifts_score.append(sum(gifts_map[i]) - sum(map(lambda x : x[i], gifts_map)))
    gifts_cnt = [0] * len(friends)
    for i in range(len(friends)):
        for j in range(len(friends)):
            if i == j:
                continue
            if (gifts_map[i][j] > gifts_map[j][i]) or (gifts_map[i][j] == gifts_map[j][i] and gifts_score[i] > gifts_score[j]):
                gifts_cnt[i] += 1
    return max(gifts_cnt)

solution(["muzi", "ryan", "frodo", "neo"], ["muzi frodo", "muzi frodo", "ryan muzi", "ryan muzi", "ryan muzi", "frodo muzi", "frodo ryan", "neo muzi"])
solution(["joy", "brad", "alessandro", "conan", "david"], ["alessandro brad", "alessandro joy", "alessandro conan", "david alessandro", "alessandro david"])
solution(["a", "b", "c"], ["a b", "b a", "c a", "a c", "a c", "c a"])
```
