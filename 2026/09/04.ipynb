```python
# [PCCP 기출문제] 1번 / 붕대 감기
def solution(bandage, health, attacks):
    cur_health = health
    cnt = 0
    for i in range(1, attacks[-1][0] + 1):
        if i == attacks[0][0]:
            cur_health -= attacks[0][1]
            attacks.pop(0)
            cnt = 0
        else:
            cur_health += bandage[1]
            cnt += 1
            if cnt == bandage[0]:
                cur_health += bandage[2]
                cnt = 0
        if cur_health <= 0:
            return -1
        cur_health = min(cur_health, health)
    return cur_health

solution([5, 1, 5], 30, [[2, 10], [9, 15], [10, 5], [11, 5]])
```

```python
# [PCCE 기출문제] 9번 / 이웃한 칸
def solution(board, h, w):
    answer = 0
    dimensions = (-1, 0), (1, 0), (0, -1), (0, 1)
    for x, y in dimensions:
        if 0 <= h + x < len(board) and 0 <= w + y < len(board[0]):
            if board[h + x][w + y] == board[h][w]:
                answer += 1
    return answer

solution([["blue", "red", "orange", "red"], ["red", "red", "blue", "orange"], ["blue", "orange", "red", "red"], ["orange", "orange", "red", "blue"]], 1, 1)
```

```python
# [PCCE 기출문제] 10번 / 데이터 분석
def solution(data, ext, val_ext, sort_by):
    filter_idx = 0
    if ext == "code":
        filter_idx = 0
    elif ext == "date":
        filter_idx = 1
    elif ext == "maximum":
        filter_idx = 2
    elif ext == "remain":
        filter_idx = 3
    
    filter_data = [d for d in data if d[filter_idx] < val_ext]
    if sort_by == "code":
        filter_data.sort(key=lambda x: x[0])
    elif sort_by == "date":
        filter_data.sort(key=lambda x: x[1])
    elif sort_by == "maximum":
        filter_data.sort(key=lambda x: x[2])
    elif sort_by == "remain":
        filter_data.sort(key=lambda x: x[3])
    
    return filter_data

solution([[1, 20300104, 100, 80], [2, 20300804, 847, 37], [3, 20300401, 10, 8]], "date", 20300501, "remain")
```

```python
# 달리기 경주
def solution(players, callings):
    players_dict = {player: i for i, player in enumerate(players)}
    for calling in callings:
        idx = players_dict[calling]
        players[idx], players[idx - 1] = players[idx - 1], players[idx]
        players_dict[players[idx]] = idx
        players_dict[players[idx - 1]] = idx - 1
    return players

solution(["mumu", "soe", "poe", "kai", "mine"], ["kai", "kai", "mine", "mine"])
```

```python
# 추억 점수
def solution(name, yearning, photo):
    missing_dict = dict()
    for i in range(len(name)):
        missing_dict[name[i]] = yearning[i]
    missing_score = []
    for i in range(len(photo)):
        missing_score.append(sum([missing_dict.get(j, 0) for j in photo[i]]))
    return missing_score

solution(["may", "kein", "kain", "radi"], [5, 10, 1, 3], [["may", "kein", "kain", "radi"],["may", "kein", "brin", "deny"], ["kon", "kain", "may", "coni"]])
```

```python
# 공원 산책
def solution(park, routes):
    x, y = -1, -1
    dimension_dict = {
        "E": (0, 1),
        "W": (0, -1),
        "S": (1, 0),
        "N": (-1, 0)
    }
    for i in range(len(park)):
        for j in range(len(park[i])):
            if park[i][j] == "S":
                x, y = i, j
                break
        if x != -1:
            break
    for route in routes:
        direction, distance = route.split(" ")
        distance = int(distance)
        dx, dy = dimension_dict[direction]
        dx, dy = dx * distance, dy * distance
        if not 0 <= x + dx < len(park) or not 0 <= y + dy < len(park[x]):
            continue
        if direction == "E" and "X" in park[x][y:y+dy+1]:
            continue
        elif direction == "W" and "X" in park[x][y+dy:y+1]:
            continue
        elif direction == "N" and "X" in map(lambda x: x[y], park[x+dx:x+1]):
            continue
        elif direction == "S" and "X" in map(lambda x: x[y], park[x:x+dx+1]):
            continue
        x, y = x + dx, y + dy
    return [x, y]

solution(["SOO","OOO","OOO"], ["E 2","S 2","W 1"])
solution(["SOO","OXX","OOO"], ["E 2","S 2","W 1"])
solution(["OSO","OOO","OXO","OOO"], ["E 2","S 3","W 1"])
park = ["OOO",
        "OXO",
        "OSO",
        "OOO"]
solution(park, ["N 2"]) 
```
