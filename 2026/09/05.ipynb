```python
# 바탕화면 정리
def solution(wallpaper):
    lux, luy, rdx, rdy = len(wallpaper), len(wallpaper[0]), 0, 0
    for i in range(len(wallpaper)):
        for j in range(len(wallpaper[i])):
            if wallpaper[i][j] == "#":
                lux = min(lux, i)
                luy = min(luy, j)
                rdx = max(rdx, i)
                rdy = max(rdy, j)
    return [lux, luy, rdx + 1, rdy + 1]

solution([".#...", "..#..", "...#."])
```

```python
# 덧칠하기
def solution(n, m, section:list):
    answer = 0
    while section:
        answer += 1
        start_idx = section.pop(0)
        while section and section[0] < start_idx + m:
            section.pop(0)
    return answer

solution(8, 4, [2, 3, 6])
```

```python
# 대충 만든 자판
def solution(keymap, targets):
    answer = []
    for target in targets:
        cnt = 0
        for i in target:
            min_cnt = 101
            for key in keymap:
                if key.find(i) != -1:
                    min_cnt = min(min_cnt, key.find(i) + 1)
            if min_cnt == 101:
                cnt = -1
                break
            print(f"i = {i}, min_cnt = {min_cnt}")
            cnt += min_cnt
        answer.append(cnt)
    return answer

solution(["ABACD", "BCEFD"], ["ABCD","AABB"]), solution(["AA"], ["B"]), solution(["AGZ", "BSSS"], ["ASA","BGZ"])
```

```python
# 카드 뭉치
def solution(cards1, cards2, goal):
    for i in goal:
        if cards1 and i == cards1[0]:
            cards1.pop(0)
        elif cards2 and i in cards2[0]:
            cards2.pop(0)
        else:
            return "No"
    return "Yes"
```

```python
# 둘만의 암호
def solution(s, skip, index):
    answer = ''
    for i in s:
        cnt = 0
        while cnt < index:
            new_chr_idx = (ord(i) + 1)
            if new_chr_idx > 122:
                new_chr_idx -= 26
            i = chr(new_chr_idx)
            if i not in skip:
                cnt += 1
        answer += i
    return answer
solution("aukks", "wbqd", 5)
```

```python
# 개인정보 수집 유효기간
DAYS_IN_MONTH = 28
MONTHS_IN_YEAR = 12
DAYS_IN_YEAR = DAYS_IN_MONTH * MONTHS_IN_YEAR  # 336


def to_ordinal(year, month, day):
    return DAYS_IN_YEAR * (year - 1) + DAYS_IN_MONTH * (month - 1) + day


def solution(today, terms, privacies):
    term_months = {t.split()[0]: int(t.split()[1]) for t in terms}
    today_ordinal = to_ordinal(*map(int, today.split(".")))

    answer = []
    for i, privacy in enumerate(privacies, start=1):
        date, term = privacy.split()
        expire = to_ordinal(*map(int, date.split("."))) + term_months[term] * DAYS_IN_MONTH
        if today_ordinal >= expire:
            answer.append(i)
    return answer

solution("2022.05.19", ["A 6", "B 12", "C 3"], ["2021.05.02 A", "2021.07.01 B", "2022.02.19 C", "2022.02.20 C"])
```
