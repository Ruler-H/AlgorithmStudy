```python
# 크기가 작은 부분 문자열
def solution(t, p):
    answer = 0
    for i in range(len(t) - len(p) + 1):
        print(int(t[i:i + len(p)]))
        if int(t[i:i + len(p)]) <= int(p):
            answer += 1
    return answer

solution("3141592", "271")
```

```python
# 가장 가까운 같은 글자
def solution(s):
    answer = []
    s_dict = dict()
    for i in range(len(s)):
        if s[i] in s_dict:
            answer.append(i - s_dict[s[i]])
        else:
            answer.append(-1)
        s_dict[s[i]] = i
    return answer

solution("banana")
```

```python
# 문자열 나누기
def solution(s):
    answer = 0
    s = list(map(lambda x : x, s))
    while s:
        answer += 1
        x = s.pop(0)
        x_cnt = 1
        none_x_cnt = 0
        while s and x_cnt != none_x_cnt:
            if x != s.pop(0):
                none_x_cnt += 1
            else:
                x_cnt += 1
                
    return answer

solution("banana"), solution("abracadabra")
```

```python
# 명예의 전당 (1)
import heapq

def solution(k, score):
    answer = []
    hall_of_fame = []  # 크기 k인 최소 힙
    for s in score:
        heapq.heappush(hall_of_fame, s)
        if len(hall_of_fame) > k:
            heapq.heappop(hall_of_fame)
        answer.append(hall_of_fame[0])
    return answer

solution(3,[10, 100, 20, 150, 1, 100, 200]), solution(4, [0, 300, 40, 300, 20, 70, 150, 50, 500, 1000])
solution(10, [1,2,3,4,5,6,7])
```

```python
# 기사단원의 무기
def solution(number, limit, power):
    answer = 0
    for i in range(1, number + 1):
        measure = set()
        for j in range(1, int(i ** 0.5) + 1):
            if i % j == 0:
                measure.add(j)
                measure.add(i // j)
        measure_len = len(measure)
        if measure_len > limit:
            answer += power
        else:
            answer += measure_len
    return answer

solution(5, 3, 2), solution(10, 3, 2)
```

```python
# 과일 장수
def solution(k, m, score):
    score.sort()
    answer = 0
    for i in range(len(score) - m, -1, -m):
        answer += score[i] * m
    return answer

solution(3, 4, [1, 2, 3, 1, 2, 3, 1])
```
