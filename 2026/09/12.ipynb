```python
# 푸드 파이트 대회
def solution(food):
    answer = '0'
    for i in range(len(food) - 1, 0, -1):
        answer = str(i) * (food[i] // 2) + answer
        answer += str(i) * (food[i] // 2)
    return answer

solution([1, 3, 4, 6])
```

```python
# 햄버거 만들기
def solution(ingredient):
    stack = []
    answer = 0
    for x in ingredient:
        stack.append(x)
        if stack[-4:] == [1, 2, 3, 1]:
            del stack[-4:]
            answer += 1
    return answer

solution([2, 1, 1, 2, 3, 1, 2, 3, 1]), solution([1, 3, 2, 1, 2, 1, 3, 1, 2])
# [2, 1, 1, 2, 3, 1, 2, 3, 1]
# [2, 1, 2, 3, 1]
# [2]
```

```python
# 옹알이 (2)
def solution(babbling):
    words = {"aya", "ye", "woo", "ma"}
    answer = 0
    for b in babbling:
        buf = ""
        last = None
        for ch in b:
            buf += ch
            if buf in words and buf != last:
                last = buf
                buf = ""
        if not buf:
            answer += 1
    return answer

# solution(["aya", "yee", "u", "maa"])
solution(["ayaye", "uuu", "yeye", "yemawoo", "ayaayaa"])
```
