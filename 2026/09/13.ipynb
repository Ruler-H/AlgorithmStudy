```python
# 콜라 문제
def solution(a, b, n):
    answer = 0
    while n >= a:
        colla = (n // a) * b
        answer += colla
        n = (n % a) + colla

    return answer

solution(2, 1, 20), solution(3, 1, 20)
```

```python
# 삼총사
def solution(number:list):
    answer = 0
    number.sort()
    for i in range(len(number) - 2):
        for j in range(i + 1, len(number) - 1):
            for k in range(j + 1, len(number)):
                num_sum = number[i] + number[j] + number[k]
                if num_sum == 0:
                    answer += 1
                elif num_sum > 0:
                    break
    return answer
solution([-2, 3, 0, 2, -5]), solution([-3, -2, -1, 0, 1, 2, 3]), solution([-1, 1, -1, 1])
```

```python
# 숫자 짝꿍
from collections import Counter

def solution(X: str, Y: str):
    cx, cy = Counter(X), Counter(Y)
    answer = ''.join(d * min(cx[d], cy[d]) for d in '9876543210')
    if not answer:
        return "-1"
    if answer[0] == '0':
        return "0"
    return answer

solution("100", "2345"), solution("100", "203045"), solution("100", "123450"), solution("12321", "42531"), solution("5525", "1255")
solution("100", "2345"), solution("1000000","0"), solution("1"*4301, "1"*4301)
```

```python
# 성격 유형 검사하기
def solution(survey, choices):
    answer = ''
    survey_dict = {
        "R": 0,
        "T": 0,
        "C": 0,
        "F": 0,
        "J": 0,
        "M": 0,
        "A": 0,
        "N": 0
    }
    for i in range(len(survey)):
        score = choices[i] - 4
        if score > 0:
            survey_dict[survey[i][-1]] += score
        else:
            survey_dict[survey[i][0]] += abs(score)
    for i, j in [("R", "T"), ("C", "F"), ("J", "M"), ("A", "N")]:
        if survey_dict[i] >= survey_dict[j]:
            answer += i
        else:
            answer += j
    return answer

solution(["AN", "CF", "MJ", "RT", "NA"], [5, 3, 2, 7, 5]), solution(["TR", "RT", "TR"], [7, 1, 3])
```
