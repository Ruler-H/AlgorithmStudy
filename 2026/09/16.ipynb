```python
# 신고 결과 받기
def solution(id_list, report, k):
    answer = [0] * len(id_list)
    reported_dict = {id: [0, []] for id in id_list}
    for i in report:
        user, reported_user = i.split(" ")
        if user not in reported_dict[reported_user][1]:
            reported_dict[reported_user][0] += 1
            reported_dict[reported_user][1].append(user)

    for i in reported_dict:
        if reported_dict[i][0] >= k:
            for user in reported_dict[i][1]:
                answer[id_list.index(user)] += 1
    return answer

solution(["muzi", "frodo", "apeach", "neo"], ["muzi frodo","apeach frodo","frodo neo","muzi neo","apeach muzi"], 2), solution(["con", "ryan"], ["ryan con", "ryan con", "ryan con", "ryan con"], 3)
```
