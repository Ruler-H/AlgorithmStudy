```python
# 나머지가 1이 되는 수 찾기
def solution(n):
    for i in range(1, n):
        if n % i == 1:
            return i

solution(10)
```

```python
# 부족한 금액 계산하기
def solution(price, money, count):
    total_price = ((count * (count + 1)) // 2) * price
    if total_price > money:
        return total_price - money
    else:
        return 0

solution(3, 20, 4)
```

```python
# 약수의 개수와 덧셈
def solution(left, right):
    answer = 0
    for i in range(left, right + 1):
        divisors_len = get_divisors_len(i)
        if divisors_len % 2 == 0:
            answer += i
        else:
            answer -= i
    return answer

def get_divisors_len(n):
    divisors = set()
    i = 1
    while i * i <= n:
        if n % i == 0:
            divisors.add(i)
            divisors.add(n // i)
        i += 1
    return len(divisors)

solution(13, 17)
```
