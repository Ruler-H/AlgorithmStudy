```python
# 선인장 숨기기
from collections import deque

def solution(m, n, h, w, drops):
    INF = len(drops) + 1
    grid = [[INF] * n for _ in range(m)]
    for t, (r, c) in enumerate(drops, 1):
        grid[r][c] = t

    def slide_min(arr, k):
        dq, out = deque(), []
        for i, v in enumerate(arr):
            while dq and arr[dq[-1]] >= v:
                dq.pop()
            dq.append(i)
            if dq[0] <= i - k:
                dq.popleft()
            if i >= k - 1:
                out.append(arr[dq[0]])
        return out

    # 1) 가로 방향: 폭 w 구간 최솟값
    row_min = [slide_min(row, w) for row in grid]
    cols = n - w + 1

    # 2) 세로 방향: 높이 h 구간 최솟값
    col_res = [slide_min([row_min[r][c] for r in range(m)], h) for c in range(cols)]

    best, ans = -1, [0, 0]
    for r in range(m - h + 1):
        for c in range(cols):
            v = col_res[c][r]
            if v > best:  # 동점이면 위쪽 → 왼쪽 우선이므로 strict 비교
                best, ans = v, [r, c]
    return ans
```
