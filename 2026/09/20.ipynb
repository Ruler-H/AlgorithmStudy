```python
# 힌트 스테이지
def solution(cost, hint):
    n = len(cost)
    cnt = [0] * (n + 1)          # cnt[s] = s번 스테이지 힌트권 보유 개수
    best = float('inf')

    def dfs(i, total):
        nonlocal best
        # i번(0-index) 스테이지 해결: 보유 힌트권 사용 (최대 n-1개)
        total += cost[i][min(cnt[i + 1], n - 1)]
        if i == n - 1:
            best = min(best, total)
            return
        # 1) 번들을 사지 않음
        dfs(i + 1, total)
        # 2) 번들을 삼
        price, *tickets = hint[i]
        for t in tickets:
            cnt[t] += 1
        dfs(i + 1, total + price)
        for t in tickets:
            cnt[t] -= 1

    dfs(0, 0)
    return best
```
