```python
# 바이러스 파이프
from functools import lru_cache


def solution(n, infection, edges, k):
    # 타입별로 "각 정점이 속한 연결 요소"를 비트마스크로 미리 계산
    comp = []
    for t in range(1, 4):
        parent = list(range(n + 1))

        def find(x):
            while parent[x] != x:
                parent[x] = parent[parent[x]]
                x = parent[x]
            return x

        for a, b, ty in edges:
            if ty == t:
                ra, rb = find(a), find(b)
                if ra != rb:
                    parent[ra] = rb

        groups = {}
        for v in range(1, n + 1):
            r = find(v)
            groups[r] = groups.get(r, 0) | (1 << v)
        comp.append([0] + [groups[find(v)] for v in range(1, n + 1)])

    @lru_cache(maxsize=None)
    def go(mask, left):
        best = bin(mask).count("1")
        if left == 0:
            return best
        for t in range(3):
            nxt, m = mask, mask
            while m:
                low = m & -m
                nxt |= comp[t][low.bit_length() - 1]   # 그 정점의 연결 요소 전체 감염
                m ^= low
            if nxt != mask:                            # 변화 없으면 조작 낭비이므로 스킵
                best = max(best, go(nxt, left - 1))
        return best

    answer = go(1 << infection, k)
    go.cache_clear()
    return answer
```
