```python
# 리프 노드 수 최대화
def solution(dist_limit, split_limit):
    def best_for(cs):
        m = len(cs)
        W = [0] * (m + 1)  # 자리 하나를 꽉 채운 서브트리의 분배 노드 수
        G = [0] * (m + 1)  # 그 서브트리가 늘려주는 리프 수
        for p in range(m - 1, -1, -1):
            W[p] = 1 + cs[p] * W[p + 1]
            G[p] = (cs[p] - 1) + cs[p] * G[p + 1]

        leaves, p, slots, budget = 1, 0, 1, dist_limit
        while p < m and budget > 0:
            if slots * W[p] <= budget:        # 이 층 전체를 꽉 채울 수 있음
                leaves += slots * G[p]
                break
            full = budget // W[p]             # 꽉 찬 서브트리 몇 개
            leaves += full * G[p]
            budget -= full * W[p]
            if budget == 0:
                break
            leaves += cs[p] - 1               # 남은 예산으로 부분 서브트리 하나
            budget -= 1
            slots = cs[p]
            p += 1
        return leaves

    answer = 1
    two, a = 1, 0
    while two <= split_limit:
        prod, b = two, 0
        while prod <= split_limit:
            answer = max(answer, best_for([2] * a + [3] * b))
            prod *= 3
            b += 1
        two *= 2
        a += 1
    return answer

solution(3, 6)
```
