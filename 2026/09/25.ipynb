```python
# 리프 노드 수 최대화
def solution(info, n, m):
    INF = float('inf')
    # dp[j] = B의 흔적 합이 j일 때 A의 흔적 합의 최솟값
    dp = [INF] * m
    dp[0] = 0

    for a, b in info:
        new = [INF] * m
        for j in range(m):
            if dp[j] == INF:
                continue
            # A가 훔치는 경우: A의 흔적이 n 미만이어야 함
            if dp[j] + a < n:
                new[j] = min(new[j], dp[j] + a)
            # B가 훔치는 경우: B의 흔적이 m 미만이어야 함
            if j + b < m:
                new[j + b] = min(new[j + b], dp[j])
        dp = new

    answer = min(dp)
    return answer if answer != INF else -1
```
