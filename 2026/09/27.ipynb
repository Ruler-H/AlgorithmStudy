```python
# 지게차와 크레인
from collections import deque

DIRS = [(-1, 0), (1, 0), (0, -1), (0, 1)]

def solution(storage, requests):
    grid = [list(row) for row in storage]
    for request in requests:
        if len(request) == 1:
            unload_by_forklift(grid, request)
        else:
            unload_by_crane(grid, request[0])
    return sum(ch != " " for row in grid for ch in row)

def unload_by_forklift(grid, target):
    n, m = len(grid), len(grid[0])
    queue = deque()
    visited = set()
    to_remove = set()

    # 가장자리 칸에서 출발
    for r in range(n):
        for c in range(m):
            if r in (0, n - 1) or c in (0, m - 1):
                visit(grid, r, c, target, queue, visited, to_remove)

    # 빈칸을 따라 안쪽으로 확장
    while queue:
        r, c = queue.popleft()
        for dr, dc in DIRS:
            nr, nc = r + dr, c + dc
            if 0 <= nr < n and 0 <= nc < m:
                visit(grid, nr, nc, target, queue, visited, to_remove)

    for r, c in to_remove:
        grid[r][c] = " "

def visit(grid, r, c, target, queue, visited, to_remove):
    if (r, c) in visited:
        return
    visited.add((r, c))
    if grid[r][c] == " ":
        queue.append((r, c))      # 빈칸이면 계속 탐색
    elif grid[r][c] == target:
        to_remove.add((r, c))     # target이면 수집만 (더 들어가지 않음)

def unload_by_crane(grid, target):
    for row in grid:
        for c in range(len(row)):
            if row[c] == target:
                row[c] = " "


solution(["AZWQY", "CAABX", "BBDDA", "ACACA"], ["A", "BB", "A"])
```
