```python
# [PCCP 기출문제] 2번 / 퍼즐 게임 챌린지
def solution(diffs, times, limit):
    def can_solve(level):
        total = times[0]
        for i in range(1, len(diffs)):
            if diffs[i] > level:
                total += (times[i] + times[i - 1]) * (diffs[i] - level) + times[i]
            else:
                total += times[i]
            if total > limit:      # 조기 종료
                return False
        return True

    left, right = 1, max(diffs)
    while left < right:
        mid = (left + right) // 2
        if can_solve(mid):
            right = mid            # 가능하면 더 낮은 레벨 시도
        else:
            left = mid + 1         # 불가능하면 레벨 올리기
    return left


# solution([1, 5, 3], [2, 4, 7], 30), solution([1, 4, 4, 2], [6, 3, 8, 2], 59)
# solution([1, 328, 467, 209, 54], [2, 7, 1, 4, 3], 1723)
solution([1, 99999, 100000, 99995], [9999, 9001, 9999, 9001], 3456789012)
```
