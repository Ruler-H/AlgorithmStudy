```python
# [PCCP 기출문제] 3번 / 충돌위험 찾기
from collections import Counter


def solution(points, routes):
    answer = 0
    robots = []
    for route in routes:
        route_idx = []
        for i in range(len(route)):
            route_idx.append(points[route[i] - 1][:])
        robots.append(route_idx)
    while robots:
        positions = [tuple(robot[0]) for robot in robots]
        cnt = Counter(positions)
        answer += sum(1 for v in cnt.values() if v >= 2)
        for i in range(len(robots) - 1, -1, -1):
            if robots[i][0] == robots[i][1]:
                robots[i].pop(1)
            if len(robots[i]) == 1:
                robots.pop(i)
        for i in range(len(robots)):
            if robots[i][0][0] > robots[i][1][0]:
                robots[i][0][0] -= 1
            elif robots[i][0][0] < robots[i][1][0]:
                robots[i][0][0] += 1
            elif robots[i][0][1] > robots[i][1][1]:
                robots[i][0][1] -= 1
            elif robots[i][0][1] < robots[i][1][1]:
                robots[i][0][1] += 1
        positions = [tuple(robot[0]) for robot in robots]
        
    return answer

# solution([[3, 2], [6, 4], [4, 7], [1, 4]], [[4, 2], [1, 3], [2, 4]])
# solution([[3, 2], [6, 4], [4, 7], [1, 4]], [[4, 2], [1, 3], [4, 2], [4, 3]])
solution([[2, 2], [2, 3], [2, 7], [6, 6], [5, 2]], [[2, 3, 4, 5], [1, 3, 4, 5]])
```
