```python
# 도넛과 막대 그래프
def solution(edges):
    out_edges, in_edges = check_in_and_out_edges(edges)
    creation_vertex, total_graph_cnt = find_creation_vertex_and_total_graph_cnt(edges, out_edges, in_edges)
    donut_cnt = 0
    bar_cnt = 0
    eight_cnt = 0
    for start in out_edges[creation_vertex]:
        v = start
        while True:
            if len(out_edges[v]) == 0:
                bar_cnt += 1
                break
            if len(out_edges[v]) == 2:
                eight_cnt += 1
                break
            v = out_edges[v][0]
            if v == start:
                donut_cnt += 1
                break
        
    return [creation_vertex, donut_cnt, bar_cnt, total_graph_cnt - donut_cnt - bar_cnt]

# 나가는 간선과 들어오는 간선을 확인해 반환하는 함수
def check_in_and_out_edges(edges):
    max_vertex = max(max(edge) for edge in edges)
    # 정점 마다 나가는 간선과 들어오는 간선
    out_edges = [[] for _ in range(max_vertex + 1)]
    in_edges = [[] for _ in range(max_vertex + 1)]
    for edge in edges:
        out_edges[edge[0]].append(edge[1])
        in_edges[edge[1]].append(edge[0])
    return out_edges, in_edges

# 생성 정점과 총 그래프 수를 반환하는 함수
def find_creation_vertex_and_total_graph_cnt(edges, out_edges, in_edges):
    for i in range(len(edges)):
        if len(out_edges[i]) > 1 and len(in_edges[i]) == 0:
            return i, len(out_edges[i])

# solution([[2, 3], [4, 3], [1, 1], [2, 1]])
solution([[4, 11], [1, 12], [8, 3], [12, 7], [4, 2], [7, 11], [4, 8], [9, 6], [10, 11], [6, 10], [3, 5], [11, 1], [5, 3], [11, 9], [3, 8]])
```
