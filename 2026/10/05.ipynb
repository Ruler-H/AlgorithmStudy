```python
# [PCCP 기출문제] 3번 / 아날로그 시계
def solution(h1, m1, s1, h2, m2, s2):
    answer = 0
    start = h1 * 3600 + m1 * 60 + s1
    end = h2 * 3600 + m2 * 60 + s2
    second, minute, hour = pos(start)
    # 시작 지점이 겹치는 경우
    answer = 1 if second == minute or second == hour else 0
    for t in range(start, end):
        second, minute, hour = pos(t)
        next_second, next_minute, next_hour = (p if p else 43200 for p in pos(t + 1))
        hit_minute = second < minute and next_second >= next_minute
        hit_hour = second < hour and next_second >= next_hour
        if hit_minute and hit_hour and next_minute == next_hour:
            answer += 1
        else:
            answer += hit_minute + hit_hour
    return answer

# (초침, 분침, 시침)의 위치를 반환하는 함수
def pos(time):
    return time % 60 * 720, time % 3600 * 12, time % 43200

solution(0, 5, 30, 0, 7, 0)
```
