```python
# 요격 시스템
def solution(targets):
    answer = 0
    last = -1  # 마지막 요격 위치 (e 직전)
    for s, e in sorted(targets, key=lambda x: x[1]):
        if s >= last:   # 이전 미사일로 못 맞힘
            answer += 1
            last = e
    return answer

solution([[4,5],[4,8],[10,14],[11,13],[5,12],[3,7],[1,4]])
```
