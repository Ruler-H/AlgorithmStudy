```python
# 두 원 사이의 정수 쌍
from math import ceil, floor, sqrt


def solution(r1, r2):
    answer = 0
    for i in range(1, r2 + 1):
        y_max = floor(sqrt((r2 ** 2) - (i ** 2)))
        if abs(i) >= r1:
            y_min = 0
        else:
            y_min = ceil(sqrt(r1 ** 2 - i ** 2))
        answer += y_max - y_min + 1
    return answer * 4

solution(2, 3)
```
