```python
# 연속된 부분 수열의 합
def solution(sequence:list, k:int):
    start = 0
    sum_value = 0
    answer = [0, len(sequence)]
    for end in range(len(sequence)):
        sum_value += sequence[end]
        while sum_value > k:
            sum_value -= sequence[start]
            start += 1
        if sum_value == k and end - start < answer[1] - answer[0]:
            answer = [start, end]
    return answer
```
