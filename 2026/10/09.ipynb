```python
# 과제 진행하기
def solution(plans:list):
    answer = []
    plans.sort(key = lambda x : x[1])
    for i in range(len(plans)):
        plans[i][1] = [int(plans[i][1][:2]), int(plans[i][1][3:])]
        plans[i][2] = int(plans[i][2])
    rest_subject = []
    cur_time = plans[0][1]
    while plans or rest_subject:
        # 현재 시각 기준 진행해야하는 계획 상 과제가 있을 경우
        if plans and plans[0][1] == cur_time:
            subject, start_time, time = plans.pop(0)
            end_time_minute = start_time[1] + time
            end_time = [start_time[0] + end_time_minute // 60, end_time_minute % 60]
            # 계획 상 마지막 과제인 경우
            if not plans:
                answer.append(subject)
            # 아직 다음 과제가 남은 경우
            else:
                # 진행하는 과제의 종료 시각이 다음 과제 시작 시간보다 늦을 경우
                next_start_time = plans[0][1]
                if end_time > next_start_time:
                    rest_subject.append([subject, diff_in_minute(next_start_time, end_time)])
                    cur_time = plans[0][1]
                else:
                    answer.append(subject)
                    cur_time = end_time
        # 현재 시각 기준 진행해야하는 계획 상 과제가 없을 경우
        else:
            # rest_subject가 있으면 가능한 시간만큼 잔여 과제 진행
            if rest_subject:
                # 계획 상 다음 과제가 있을 경우
                if plans:
                    available_time = diff_in_minute(plans[0][1], cur_time)
                    if available_time >= rest_subject[-1][1]:
                        subject, rest_time = rest_subject.pop()
                        answer.append(subject)
                        cur_time = [cur_time[0] + (cur_time[1] + rest_time) // 60, (cur_time[1] + rest_time) % 60]
                    else:
                        rest_subject[-1][1] -= available_time
                        cur_time = [cur_time[0] + (cur_time[1] + available_time) // 60, (cur_time[1] + available_time) % 60]
                # 계획 상 다음 과제가 없을 경우
                else:
                    while rest_subject:
                        subject, _ = rest_subject.pop()
                        answer.append(subject)
            # rest_subject도 없으면 다음 과제 시간으로 이동
            else:
                cur_time = plans[0][1]

    return answer

def diff_in_minute(start, end):
    return abs((end[0] * 60 + end[1]) - (start[0] * 60 + start[1]))

# solution([["korean", "11:40", "30"], ["english", "12:10", "20"], ["math", "12:30", "40"]]) # ["korean", "english", "math"]
# solution([["science", "12:40", "50"], ["music", "12:20", "40"], ["history", "14:00", "30"], ["computer", "12:30", "100"]]) # ["science", "history", "computer", "music"]
solution([["aaa", "12:00", "20"], ["bbb", "12:10", "30"], ["ccc", "12:40", "10"]]) # ["bbb", "ccc", "aaa"]
```
