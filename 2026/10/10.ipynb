```python
# 광물 캐기
def solution(picks, minerals):
    results = []
    for i in range(len(picks)):
        if picks[i]:
            if i == 0:
                results.append([len(minerals[:5]), [picks[0] - 1] + picks[1:]])
            elif i == 1:
                fatigue = 0
                for mineral in minerals[:5]:
                    if mineral == "diamond":
                        fatigue += 5
                    else:
                        fatigue += 1
                results.append([fatigue, [picks[0], picks[1] - 1, picks[2]]])
            else:
                fatigue = 0
                for mineral in minerals[:5]:
                    if mineral == "diamond":
                        fatigue += 25
                    elif mineral == "iron":
                        fatigue += 5
                    else:
                        fatigue += 1
                results.append([fatigue, picks[:2] + [picks[2] - 1]])
    minerals = minerals[5:]
    while minerals:
        temp_results = []
        for temp_fatigue, temp_picks in results:
            for i in range(len(temp_picks)):
                if temp_picks[i]:
                    if i == 0:
                        temp_results.append([temp_fatigue + len(minerals[:5]), [temp_picks[0] - 1] + temp_picks[1:]])
                    elif i == 1:
                        fatigue = 0
                        for mineral in minerals[:5]:
                            if mineral == "diamond":
                                fatigue += 5
                            else:
                                fatigue += 1
                        temp_results.append([temp_fatigue + fatigue, [temp_picks[0], temp_picks[1] - 1, temp_picks[2]]])
                    else:
                        fatigue = 0
                        for mineral in minerals[:5]:
                            if mineral == "diamond":
                                fatigue += 25
                            elif mineral == "iron":
                                fatigue += 5
                            else:
                                fatigue += 1
                        temp_results.append([temp_fatigue + fatigue, temp_picks[:2] + [temp_picks[2] - 1]])
        minerals = minerals[5:]
        if temp_results:
            results = temp_results
    return min(results, key=lambda x : x[0])[0]

# solution([1, 3, 2], ["diamond", "diamond", "diamond", "iron", "iron", "diamond", "iron", "stone"])
solution([0, 1, 1], ["diamond", "diamond", "diamond", "diamond", "diamond", "iron", "iron", "iron", "iron", "iron", "diamond"])
```
