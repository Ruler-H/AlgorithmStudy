```python
# 리코쳇 로봇
def solution(board):
    visited = [[False] * len(i) for i in board]
    R_positions = []
    for i in range(len(board)):
        for j in range(len(board[i])):
            if board[i][j] == "R":
                visited[i][j] = True
                R_positions.append((i, j))
                break
        if R_positions:
            break
    count = 0
    while R_positions:
        next_R_positions = []
        for i, j in R_positions:
            for dx, dy in [(1, 0), (-1, 0), (0, 1), (0, -1)]:
                nx, ny = i, j
                while 0 <= nx + dx < len(board) and 0 <= ny + dy < len(board[0]) and board[nx + dx][ny + dy] != "D":
                    nx, ny = nx + dx, ny + dy
                if board[nx][ny] == "G":
                    return count + 1
                if not visited[nx][ny]:
                    visited[nx][ny] = True
                    next_R_positions.append((nx, ny))
        R_positions = next_R_positions
        count += 1
    return -1
solution(["...D..R", ".D.G...", "....D.D", "D....D.", "..D...."]) # 7
```
