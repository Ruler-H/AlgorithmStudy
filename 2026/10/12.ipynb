```python
# 당구 연습
def solution(m, n, startX, startY, balls):
    answer = []
    for x, y in balls:
        cands = []
        # 왼쪽 벽 (x=0): 같은 y줄에서 공이 왼쪽에 있으면 불가
        if not (y == startY and x < startX):
            cands.append((startX + x) ** 2 + (startY - y) ** 2)
        # 오른쪽 벽 (x=m)
        if not (y == startY and x > startX):
            cands.append((2 * m - startX - x) ** 2 + (startY - y) ** 2)
        # 아래쪽 벽 (y=0)
        if not (x == startX and y < startY):
            cands.append((startX - x) ** 2 + (startY + y) ** 2)
        # 위쪽 벽 (y=n)
        if not (x == startX and y > startY):
            cands.append((startX - x) ** 2 + (2 * n - startY - y) ** 2)
        answer.append(min(cands))
    return answer

solution(10, 10, 3, 7, [[7, 7], [2, 7], [7, 3]])
```
