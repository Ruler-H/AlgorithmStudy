```python
# 미로 탈출
from collections import deque

def solution(maps):
    answer = 0
    start_x, start_y = None, None
    for i in range(len(maps)):
        for j in range(len(maps[i])):
            if maps[i][j] == "S":
                start_x, start_y = i, j
                break
        if start_x != None or start_y != None:
            break
    dxy = [(0, 1), (0, -1), (1, 0), (-1, 0)]

    # start에서 target 문자까지의 최단 거리와 도착 좌표 반환 (도달 불가 시 -1)
    def bfs(start_x, start_y, target):
        dist = [[-1] * len(maps[0]) for _ in range(len(maps))]
        dist[start_x][start_y] = 0
        queue = deque([(start_x, start_y)])
        while queue:
            x, y = queue.popleft()
            if maps[x][y] == target:
                return dist[x][y], x, y
            for dx, dy in dxy:
                nx, ny = x + dx, y + dy
                if 0 <= nx < len(maps) and 0 <= ny < len(maps[0]) and maps[nx][ny] != "X" and dist[nx][ny] == -1:
                    dist[nx][ny] = dist[x][y] + 1
                    queue.append((nx, ny))
        return -1, None, None

    # 스위치(레버)까지 이동
    dist_to_switch, switch_x, switch_y = bfs(start_x, start_y, "L")
    if dist_to_switch == -1:
        return -1

    # 스위치에서 출구까지 이동
    dist_to_exit, _, _ = bfs(switch_x, switch_y, "E")
    if dist_to_exit == -1:
        return -1

    answer = dist_to_switch + dist_to_exit
    return answer

solution(["SOOOL","XXXXO","OOOOO","OXXXX","OOOOE"]) # 16
```
