```python
# 뒤에 있는 큰 수 찾기
def solution(numbers):
    answer = [-1] * len(numbers)
    stack = []
    for i, num in enumerate(numbers):
        while stack and numbers[stack[-1]] < num:
            answer[stack.pop()] = num
        stack.append(i)
    return answer

# solution([2, 3, 3, 5]) # [3, 5, 5, -1]
solution([9, 1, 5, 3, 6, 2]) # [-1, 5, 6, 6, -1, -1]
```
