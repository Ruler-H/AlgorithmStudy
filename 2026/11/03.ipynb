```python
# 숫자 변환하기
def solution(x, y, n):
    answer = 0
    if x == y:
        return answer
    x_set = set([x])
    while x_set:
        imsi_set = set()
        for x in x_set:
            if (x + n) <= y:
                imsi_set.add(x + n)
            if (x * 2) <= y:
                imsi_set.add(x * 2)
            if (x * 3) <= y:
                imsi_set.add(x * 3)
        x_set = imsi_set
        answer += 1
        if y in x_set:
            return answer
    return -1

solution(10, 40, 5) # 2
# solution(10, 40, 30) # 1
# solution(2, 5, 4) # -1
```
