```python
# 이모티콘 할인행사
from itertools import product


def solution(users, emoticons):
    answer = [0, 0]
    for rates in product((10, 20, 30, 40), repeat=len(emoticons)):
        plus, revenue = 0, 0
        for want_rate, want_cost in users:
            total = sum(price * (100 - r) // 100 for price, r in zip(emoticons, rates) if r >= want_rate)
            if total >= want_cost:
                plus += 1
            else:
                revenue += total
        answer = max(answer, [plus, revenue])
    return answer

# solution([[40, 10000], [25, 10000]], [7000, 9000]) # [1, 5400]
solution([[40, 2900], [23, 10000], [11, 5200], [5, 5900], [40, 3100], [27, 9200], [32, 6900]], [1300, 1500, 1600, 4900]) # [4, 13860]
```
