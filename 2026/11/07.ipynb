```python
# 마법의 엘리베이터
def solution(storey):
    answer = 0
    while storey:
        d = storey % 10
        nxt = (storey // 10) % 10
        if d > 5 or (d == 5 and nxt >= 5):
            answer += 10 - d
            storey = storey // 10 + 1
        else:
            answer += d
            storey //= 10
    return answer

solution(99)
```
