```python
# 유사 칸토어 비트열
def solution(n, l, r):
    def count(n, x):
        if n == 0:
            return x
        size, ones = 5 ** (n - 1), 4 ** (n - 1)
        q, rem = divmod(x, size)
        res = (q - 1 if q >= 3 else q) * ones
        if q != 2:
            res += count(n - 1, rem)
        return res
    return count(n, r) - count(n, l - 1)

solution(10, 4, 17)
```
