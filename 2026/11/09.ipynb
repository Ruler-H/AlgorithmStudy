```python
# 테이블 해시 함수
def solution(data:list, col, row_begin, row_end):
    answer = 0
    data.sort(key=lambda x:(x[col - 1], -x[0]))
    for i in range(row_begin, row_end + 1):
        s_i = sum(v % i for v in data[i - 1])
        answer ^= s_i
    return answer

solution([[2,2,6],[1,5,10],[4,2,9],[3,8,3]], 2, 2, 3)
```
