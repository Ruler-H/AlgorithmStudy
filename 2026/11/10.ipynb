```python
import heapq

# 디펜스 게임
def solution(n, k, enemy):
    heap = []
    for i, e in enumerate(enemy):
        heapq.heappush(heap, -e)
        n -= e
        if n < 0:
            if k == 0:
                return i
            n += -heapq.heappop(heap)
            k -= 1
    return len(enemy)


solution(7, 3, [4, 2, 4, 5, 3, 3, 1]), solution(2, 4, [3, 3, 3, 3])
```
