```python
# 귤 고르기
def solution(k, tangerine):
    answer = 0
    tangerine_dict = dict()
    for i in tangerine:
        if i not in tangerine_dict:
            tangerine_dict[i] = 1
        else:
            tangerine_dict[i] += 1
    tangerine_cnt = 0
    for key, value in sorted(tangerine_dict.items(), key=lambda x: -x[1]):
        answer += 1
        tangerine_cnt += value
        if tangerine_cnt >= k:
            break
    return answer

solution(6, [1, 3, 2, 5, 4, 5, 2, 3]) # 3
# solution(4, [1, 3, 2, 5, 4, 5, 2, 3]) # 2
# solution(2, [1, 1, 1, 1, 2, 2, 2, 3]) # 1
```
