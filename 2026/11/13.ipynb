```python
import math

# 숫자 카드 나누기
def solution(arrayA, arrayB):
    a = check(math.gcd(*arrayA), arrayB)
    b = check(math.gcd(*arrayB), arrayA)
    return max(a, b)
    
def check(g, other):
    # g로 other의 어떤 수도 나눌 수 없으면 g, 아니면 0
    return 0 if any(x % g == 0 for x in other) else g

solution([10, 17], [5, 20]), solution([10, 20], [5, 17]), solution([14, 35, 119], [18, 30, 102])
```
