```python
# 우박수열 정적분
def solution(k, ranges):
    # 1. 우박수열 생성
    seq = [k]
    while k > 1:
        k = k // 2 if k % 2 == 0 else k * 3 + 1
        seq.append(k)
    n = len(seq) - 1  # x축 마지막 좌표

    # 2. 구간별 사다리꼴 넓이의 누적합
    prefix = [0.0]
    for i in range(n):
        prefix.append(prefix[-1] + (seq[i] + seq[i + 1]) / 2)

    # 3. 각 구간 [a, n+b]의 정적분
    answer = []
    for a, b in ranges:
        end = n + b
        answer.append(-1.0 if a > end else prefix[end] - prefix[a])
    return answer
```
