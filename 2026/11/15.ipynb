```python
# 롤케이크 자르기
from collections import Counter


def solution(topping):
    right = Counter(topping)
    left = set()
    answer = 0
    for t in topping:
        left.add(t)
        right[t] -= 1
        if right[t] == 0:
            del right[t]
        if len(left) == len(right):
            answer += 1
    return answer

solution([1, 2, 1, 3, 1, 4, 1, 2]), solution([1, 2, 3, 1, 4]) # 0
```
