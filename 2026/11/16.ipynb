```python
# 택배상자
def solution(order):
    n = len(order)
    assist = []
    idx = 0
    for box in range(1, n + 1):
        assist.append(box)
        while assist and assist[-1] == order[idx]:
            assist.pop()
            idx += 1
            if idx == n:
                return n
    return idx

solution([4, 3, 1, 2, 5]), solution([5, 4, 3, 2, 1]) # 5
```
