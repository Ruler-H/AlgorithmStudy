```python
# 연속 부분 수열 합의 개수
def solution(elements):
    new_elements = elements.copy() + elements.copy()
    answer_set = set()
    for i in range(1, len(elements) + 1):
        for j in range(len(elements)):
            answer_set.add(sum(new_elements[j:j+i]))
    return len(answer_set)

solution([7,9,1,1,4])
```
