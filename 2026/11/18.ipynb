```python
# 혼자 놀기의 달인
def solution(cards):
    visited = [False] * len(cards)
    result = []
    for i in range(len(cards)):
        if not visited[i]:
            result.append(grouping(visited, i, cards, []))
    if len(result) < 2:
        return 0
    answer = 1
    result.sort(reverse=True)
    return result[0] * result[1]

def grouping(visited, cur, cards, group):
    if visited[cur]:
        return len(group)
    visited[cur] = True
    group.append(cur)
    return grouping(visited, cards[cur] - 1, cards, group)

solution([8,6,3,7,2,5,1,4])
```
