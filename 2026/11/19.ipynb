```python
# 할인 행사
from collections import Counter


def solution(want, number, discount):
    want_counter = Counter(dict([(w, n) for w, n in zip(want, number)]))
    purchase_counter = Counter(discount[:9])
    for i in range(9, len(discount)):
        purchase_counter[discount[i]] += 1
        if want_counter == purchase_counter:
            return i - 8
        purchase_counter[discount[i - 9]] -= 1
    return 0

solution(["banana", "apple", "rice", "pork", "pot"], [3, 2, 2, 2, 1], ["chicken", "apple", "apple", "banana", "rice", "apple", "pork", "banana", "pork", "rice", "pot", "banana", "apple", "banana"])
solution(["apple"], [10], ["banana", "banana", "banana", "banana", "banana", "banana", "banana", "banana", "banana", "banana"])
```
