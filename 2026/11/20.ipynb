```python
from collections import deque

# 두 큐 합 같게 만들기
def solution(queue1, queue2):
    q1, q2 = deque(queue1), deque(queue2)
    s1 = sum(q1)
    s2 = sum(q2)
    if (s1 + s2) % 2:
        return -1
    for i in range(len(queue1) * 3 + 1):
        if s1 > s2:
            num = q1.popleft()
            q2.append(num)
            s1 -= num
            s2 += num
        elif s2 > s1:
            num = q2.popleft()
            q1.append(num)
            s2 -= num
            s1 += num
        else:
            return i
    return -1

solution([3, 2, 7, 2], [4, 6, 5, 1]) # 2
solution([1, 2, 1, 2], [1, 10, 1, 2]) # 7
solution([1, 1], [1, 5])
```
