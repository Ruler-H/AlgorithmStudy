```python
# 양궁대회
def solution(n, info):
    best_diff, answer = 0, [-1]
    for mask in range(1 << 10):          # 10점~1점 중 라이언이 가져갈 점수 조합
        ryan = [0] * 11
        used = 0
        lion = apeach = 0
        for i in range(10):
            if mask & (1 << i):
                ryan[i] = info[i] + 1    # 어피치보다 딱 1발 더
                used += ryan[i]
                lion += 10 - i
            elif info[i] > 0:
                apeach += 10 - i
        if used > n:
            continue
        ryan[10] = n - used               # 남은 화살은 0점에
        diff = lion - apeach
        if diff <= 0:
            continue
        # 점수 차가 더 크거나, 같으면 낮은 점수를 더 많이 맞힌 쪽 우선
        if diff > best_diff or (diff == best_diff and ryan[::-1] > answer[::-1]):
            best_diff, answer = diff, ryan
    return answer
```
