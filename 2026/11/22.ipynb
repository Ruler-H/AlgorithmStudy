```python
# 주차 요금 계산
import math
from collections import defaultdict

def solution(fees, records):
    base_time, base_fee, unit_time, unit_fee = fees
    in_time = {}                 # 현재 주차 중인 차량의 입차 시각(분)
    total = defaultdict(int)     # 차량별 누적 주차 시간(분)

    for r in records:
        t, car, kind = r.split()
        h, m = map(int, t.split(':'))
        minute = h * 60 + m
        if kind == 'IN':
            in_time[car] = minute
        else:
            total[car] += minute - in_time.pop(car)

    # 출차 기록이 없으면 23:59에 출차한 것으로 처리
    for car, t in in_time.items():
        total[car] += 23 * 60 + 59 - t

    answer = []
    for car in sorted(total):    # 차량 번호가 작은 순
        t = total[car]
        fee = base_fee
        if t > base_time:
            fee += math.ceil((t - base_time) / unit_time) * unit_fee
        answer.append(fee)
    return answer
```
