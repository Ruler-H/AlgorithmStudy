```python
# k진수에서 소수 개수 구하기
def solution(n, k):
    # 1. n을 k진수 문자열로 바꾸기
    digits = ''
    while n > 0:
        digits = str(n % k) + digits
        n //= k

    # 2. '0'을 기준으로 나누고, 빈 문자열이 아닌 조각만 소수인지 확인
    answer = 0
    for part in digits.split('0'):
        if part and is_prime(int(part)):
            answer += 1
    return answer

# k진수에서 소수 개수 구하기
def is_prime(x):
    if x < 2:
        return False
    i = 2
    while i * i <= x:
        if x % i == 0:
            return False
        i += 1
    return True
```
