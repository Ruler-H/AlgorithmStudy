```python
# 피로도
from itertools import permutations


def solution(k, dungeons):
    answer = 0
    for order in permutations(dungeons):
        cur, cnt = k, 0
        for need, cost in order:
            if cur >= need:
                cur -= cost
                cnt += 1
        answer = max(answer, cnt)
    return answer

solution(80, [[80,20],[50,40],[30,10]]) # 3
```
