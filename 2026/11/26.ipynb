```python
# 교점에 별 만들기
def solution(line):
    # 교점 구하기
    meet = []
    for i in range(len(line)):
        A, B, E = line[i]
        for j in range(i + 1, len(line)):
            C, D, F = line[j]
            if not (A * D - B * C) == 0:
                x = (B * F - E * D) / (A * D - B * C)
                y = (E * C - A * F) / (A * D - B * C)
                if x % 1 == 0 and y % 1 == 0:
                    meet.append((int(x), int(y)))

    # 격자 크기 정하기
    max_x = max(meet, key=lambda x: x[0])[0]
    min_x = min(meet, key=lambda x: x[0])[0]
    max_y = max(meet, key=lambda x: x[1])[1]
    min_y = min(meet, key=lambda x: x[1])[1]
    board = [["."] * (max_x - min_x + 1) for i in range(max_y - min_y + 1)]
    
    # 교점 찍기
    for x, y in meet:
        x = x - min_x
        y = -y + max_y
        board[y][x] = "*"
    
    return ["".join(row) for row in board]

solution([[2, -1, 4], [-2, -1, 4], [0, -1, 1], [5, -8, -12], [5, 8, 12]])
```
