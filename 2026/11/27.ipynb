```python
# 전력망을 둘로 나누기
from collections import defaultdict, deque

def solution(n, wires):
    graph = defaultdict(list)
    for a, b in wires:
        graph[a].append(b)
        graph[b].append(a)

    def count(start, cut_a, cut_b):
        visited = {start}
        q = deque([start])
        while q:
            cur = q.popleft()
            for nxt in graph[cur]:
                # 끊은 전선은 건너뜀
                if (cur == cut_a and nxt == cut_b) or (cur == cut_b and nxt == cut_a):
                    continue
                if nxt not in visited:
                    visited.add(nxt)
                    q.append(nxt)
        return len(visited)

    answer = n
    for a, b in wires:
        cnt = count(a, a, b)              # 전선 a-b를 끊었을 때 a 쪽 송전탑 개수
        answer = min(answer, abs(n - 2 * cnt))  # |cnt - (n - cnt)|
    return answer
```
