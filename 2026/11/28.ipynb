```python
# 빛의 경로 사이클
def solution(grid):
    n, m = len(grid), len(grid[0])
    # 방향: 0=상, 1=우, 2=하, 3=좌
    dr = [-1, 0, 1, 0]
    dc = [0, 1, 0, -1]
    visited = [[[False] * 4 for _ in range(m)] for _ in range(n)]
    answer = []

    for i in range(n):
        for j in range(m):
            for d in range(4):
                if visited[i][j][d]:
                    continue
                r, c, dir_ = i, j, d
                length = 0
                while not visited[r][c][dir_]:
                    visited[r][c][dir_] = True
                    length += 1
                    # 현재 칸에서 dir_ 방향으로 나가서 다음 칸에 도착
                    r = (r + dr[dir_]) % n
                    c = (c + dc[dir_]) % m
                    # 도착한 칸의 문자에 따라 방향 전환
                    if grid[r][c] == 'L':
                        dir_ = (dir_ + 3) % 4
                    elif grid[r][c] == 'R':
                        dir_ = (dir_ + 1) % 4
                answer.append(length)

    return sorted(answer)
```
